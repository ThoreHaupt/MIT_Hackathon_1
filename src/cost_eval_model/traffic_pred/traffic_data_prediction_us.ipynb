```python
import tensorflow as tf
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Example DataFrame structure (replace with your data)
# df = pd.read_csv("traffic_data.csv") 

# Simulate some sample data for demonstration
np.random.seed(42)
data_size = 1000
df = pd.DataFrame({
    'day_of_week': np.random.randint(0, 7, size=data_size),
    'hour': np.random.randint(0, 24, size=data_size),
    'month': np.random.randint(0, 12, size=data_size),
    'traffic_volume': np.random.normal(loc=200, scale=50, size=data_size)  # Simulated traffic volume
})

# Features and target
X = df[['day_of_week', 'hour', 'month']].values
y = df['traffic_volume'].values

# Normalize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# train/test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Build the model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(3,)),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(1)  # Predict traffic volume
])

model.compile(optimizer='adam', loss='mse', metrics=['mae'])

# train the model
model.fit(X_train, y_train, epochs=50, validation_data=(X_test, y_test), verbose=1)

# Predict traffic volume for Tuesday at 8AM
sample_input = scaler.transform([[1, 8, 3]])  # 1 = Tuesday
predicted_volume = model.predict(sample_input)

print(f"Predicted traffic volume for Tuesday at 8AM: {predicted_volume[0][0]:.2f}")


```
